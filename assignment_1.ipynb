```python
28.
def swap_message(val1, val2):
    # Swap local variables
    val1, val2 = val2, val1
    print("Inside function:")
    print("val1 =", val1)
    print("val2 =", val2)


# Original variables
x = 6
y = 3

# Call the function
swap_message(x, y)

# Print original variables
print("\nOutside function:")
print("x =", x)
print("y =", y)
```
