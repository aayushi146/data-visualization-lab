q29

```python
inventory_df = pd.DataFrame({
    "Item": ["Pen", "Book", "Pencil", "Bag", "Scale"],
    "Quantity": [10, 5, 20, 3, 8],
    "Price": [10, 100, 5, 500, 20]
})

# Partially indexed Series
stock = pd.Series(
    [50, 80],
    index=[1, 3]
)

# Add new column
inventory_df["Stock"] = stock
print(inventory_df)

# Delete the added column
inventory_df.drop("Stock", axis=1, inplace=True)
print(inventory_df)
```

```python
import pandas as pd

# Create sample data
lab_df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Neha"],
    "Age": [19, 20, 21],
    "Total Marks": [85, 90, 78],
    "City": ["Delhi", "Agra", "Patna"]
})

# Display DataFrame
print(lab_df)

# Display column names
print(lab_df.columns)

# Retrieve one single-word column
print(lab_df["Name"])

# Retrieve column containing a space
print(lab_df["Total Marks"])

# Create smaller DataFrame
small_df = lab_df[["Name", "Total Marks", "City"]]
print(small_df)

# Add a new column
lab_df["Status"] = "Passed"
print(lab_df)

# Delete the new column
lab_df.drop("Status", axis=1, inplace=True)
print(lab_df)
```
