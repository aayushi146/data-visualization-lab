```python
import numpy as np

np.random.seed(42)

marks = np.random.randint(20, 101, size=(12, 4))

print("Marks:\n", marks)

# Average marks of each student
student_avg = np.mean(marks, axis=1)
print("\nStudent averages:\n", student_avg)

# Mean and standard deviation of each test
test_mean = np.mean(marks, axis=0)
test_std = np.std(marks, axis=0)

print("\nTest means:", test_mean)
print("Test standard deviations:", test_std)

# Students passing all four tests
passed_all = np.all(marks >= 40, axis=1)
print("\nStudents passing all tests:", np.where(passed_all)[0] + 1)
print("Number passing all tests:", np.sum(passed_all))

# Students with average >= 75
high_avg = np.where(student_avg >= 75)[0] + 1
print("\nStudents with average >= 75:", high_avg)

# Test with greatest variation
print("\nTest with greatest variation:",
      np.argmax(test_std) + 1)
```

Q36

```python
import numpy as np

# Create a 5 x 5 matrix containing 21 to 45
a = np.arange(21, 46).reshape(5, 5)
print("Original matrix:\n", a)

# Extract central 3 x 3 matrix
b = a[1:4, 1:4]
print("\nCentral matrix:\n", b)

# Transpose
c = b.T
print("\nTranspose:\n", c)

# Matrix multiplication
product = b @ c
print("\nProduct:\n", product)

# Row-wise sums
row_sums = np.sum(product, axis=1)
print("\nRow sums:", row_sums)

# Maximum and its positions
maximum = np.max(row_sums)
positions = np.where(row_sums == maximum)[0]

print("Maximum row sum:", maximum)
print("Position (zero-based):", positions)
print("Row number:", positions + 1)
```

```python
import numpy as np

quantities = np.array([
    [10, 20,  5,  8],
    [15, 18, 12, 10],
    [20, 25, 10, 15],
    [18, 22, 15, 12],
    [25, 20,  8, 20],
    [16, 24, 18, 15]
])

original_prices = np.array([150, 240, 90, 320])

# Revenue for each day
daily_revenue = quantities @ original_prices
print("Daily revenue:", daily_revenue)

# Total revenue per product
total_quantity = quantities.sum(axis=0)
product_revenue = total_quantity * original_prices

print("Total quantity per product:", total_quantity)
print("Revenue per product:", product_revenue)

# Highest-revenue day
print("Highest-revenue day:", np.argmax(daily_revenue) + 1)

# Best-selling product by quantity
print("Best-selling product:", np.argmax(total_quantity) + 1)

# Reduce prices by 10% if quantity exceeds 100
new_prices = np.where(
    total_quantity > 100,
    original_prices * 0.90,
    original_prices
)

print("Original prices:", original_prices)
print("Reduced prices:", new_prices)

# Recalculate revenue
new_daily_revenue = quantities @ new_prices
print("New total revenue:", new_daily_revenue.sum())
```

```python
import numpy as np

np.random.seed(42)

# Generate 100 normal observations
data = np.random.normal(50, 12, 100)

# Count observations outside 30 to 70
outside = (data < 30) | (data > 70)
print("Values outside range:", np.sum(outside))

# Replace values below 30 with 30 and above 70 with 70
cleaned = np.where(data < 30, 30,
                   np.where(data > 70, 70, data))

# Compare statistics
print("\nOriginal data:")
print("Mean:", np.mean(data))
print("Variance:", np.var(data))
print("Standard deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))

print("\nCleaned data:")
print("Mean:", np.mean(cleaned))
print("Variance:", np.var(cleaned))
print("Standard deviation:", np.std(cleaned))
print("Minimum:", np.min(cleaned))
print("Maximum:", np.max(cleaned))
```

```python
import pandas as pd

branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})

# Ordinary addition
ordinary = branch_a + branch_b + branch_c
print("Ordinary addition:")
print(ordinary)

# Quarters common to all branches
common = branch_a.index.intersection(branch_b.index)
common = common.intersection(branch_c.index)

print("\nCommon quarters:", list(common))

# Combined earnings for common quarters
common_total = branch_a[common] + branch_b[common] + branch_c[common]
print("\nCommon-quarter earnings:")
print(common_total)

# Treat missing branch records as zero
zero_total = (
    branch_a.add(branch_b, fill_value=0)
            .add(branch_c, fill_value=0)
)

print("\nTotal treating missing records as zero:")
print(zero_total)
```

```python
import pandas as pd

scores = {
    'R101': 78,
    'R102': 65,
    'R104': 92,
    'R105': 48
}

required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']

# Preserve the required index order
s = pd.Series(scores, index=required)

print("Original Series:")
print(s)

# Identify missing records
missing = s[s.isnull()]
print("\nMissing student records:")
print(missing)

# Filter valid scores >= 70
filtered = s[s.notnull() & (s >= 70)]
print("\nValid scores >= 70:")
print(filtered)

# Convert to dictionary
data_dict = filtered.to_dict()
print("\nDictionary:")
print(data_dict)

# Reconstruct Series
new_series = pd.Series(data_dict)
new_series = new_series.reindex(filtered.index)

print("\nReconstructed Series:")
print(new_series)

# Verify labelled values
print("\nValues match:", filtered.equals(new_series))
```
